````python
import sys
import os

# Add the zellaforge/backend directory to sys.path
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), 'zellaforge', 'backend')))

from celery import Celery
from app.engine.executor import WorkflowExecutor
import asyncio

celery = Celery(
    "worker",
    broker="redis://redis:6379/0",
    backend="redis://redis:6379/0"
)

@celery.task
def run_workflow(workflow_id: int, trigger_data: dict):
    from app.models.workflow import Workflow
    from app.core.database import SessionLocal

    db = SessionLocal()
    workflow = db.query(Workflow).filter(Workflow.id == workflow_id).first()
    if not workflow:
        return

    executor = WorkflowExecutor(workflow, trigger_data)
    asyncio.run(executor.run())
    db.close()
````

````python
import os

workflow_api_content = """from fastapi import APIRouter, Depends, HTTPException
from sqlalchemy.orm import Session
from typing import List
from ...core.database import get_db
from ...core.security import get_current_user
from ...models.workflow import Workflow
from ...schemas.workflow import WorkflowCreate, WorkflowUpdate, WorkflowResponse

router = APIRouter(prefix="/workflows", tags=["workflows"])

@router.post("/", response_model=WorkflowResponse)
def create_workflow(
    data: WorkflowCreate,
    db: Session = Depends(get_db),
    user=Depends(get_current_user)
):
    wf = Workflow(
        name=data.name,
        description=data.description,
        owner_id=user.id,
        nodes=[n.dict() for n in data.nodes],
        edges=[e.dict() for e in data.edges]
    )
    db.add(wf)
    db.commit()
    db.refresh(wf)
    return wf

@router.get("/", response_model=List[WorkflowResponse])
def list_workflows(db: Session = Depends(get_db), user=Depends(get_current_user)):
    return db.query(Workflow).filter(Workflow.owner_id == user.id).all()

@router.put("/{wf_id}", response_model=WorkflowResponse)
def update_workflow(
    wf_id: int,
    data: WorkflowUpdate,
    db: Session = Depends(get_db),
    user=Depends(get_current_user)
):
    wf = db.query(Workflow).filter(Workflow.id == wf_id, Workflow.owner_id == user.id).first()
    if not wf:
        raise HTTPException(404, "Not found")
    if data.name: wf.name = data.name
    if data.nodes: wf.nodes = [n.dict() for n in data.nodes]
    if data.edges: wf.edges = [e.dict() for e in data.edges]
    if data.is_active is not None: wf.is_active = data.is_active
    db.commit()
    db.refresh(wf)
    return wf"""

os.makedirs(os.path.dirname('zellaforge/backend/app/api/v1/workflows.py'), exist_ok=True)
with open('zellaforge/backend/app/api/v1/workflows.py', 'w') as f:
    f.write(workflow_api_content)

print("workflows.py created successfully in zellaforge/backend/app/api/v1/.")
````

````python
import os

docs_api_content = """from fastapi import APIRouter, Depends, WebSocket, WebSocketDisconnect
from sqlalchemy.orm import Session
from typing import List
from ...core.database import get_db
from ...core.security import get_current_user
from ...models.document import Document
from ...schemas.document import DocumentCreate, DocumentUpdate, DocumentResponse
from ...collab.yjs_server import YDocManager

router = APIRouter(prefix="/docs", tags=["docs"])

@router.post("/", response_model=DocumentResponse)
def create_doc(
    data: DocumentCreate,
    db: Session = Depends(get_db),
    user=Depends(get_current_user)
):
    doc = Document(**data.dict(), owner_id=user.id)
    db.add(doc)
    db.commit()
    db.refresh(doc)
    return doc

@router.get("/", response_model=List[DocumentResponse])
def list_docs(
    db: Session = Depends(get_db),
    user=Depends(get_current_user)
):
    return db.query(Document).filter(Document.owner_id == user.id).all()

@router.websocket("/ws/{doc_id}")
async def websocket_doc(websocket: WebSocket, doc_id: int, user=Depends(get_current_user)):
    await websocket.accept()
    ydoc = YDocManager.get_doc(str(doc_id))
    update = YDocManager.get_update(str(doc_id))
    await websocket.send_bytes(update)

    try:
        while True:
            data = await websocket.receive_bytes()
            YDocManager.apply_update(str(doc_id), data)
            # In prod: broadcast via Redis
            await websocket.send_bytes(data)
    except WebSocketDisconnect:
        pass"""

os.makedirs(os.path.dirname('zellaforge/backend/app/api/v1/docs.py'), exist_ok=True)
with open('zellaforge/backend/app/api/v1/docs.py', 'w') as f:
    f.write(docs_api_content)

print("docs.py created successfully in zellaforge/backend/app/api/v1/.")
````

````python
import os

auth_content = """from fastapi import APIRouter, Depends, HTTPException, status
from fastapi.security import OAuth2PasswordRequestForm
from sqlalchemy.orm import Session
from datetime import timedelta
from ...core.database import get_db
from ...core.security import authenticate_user, create_token, hash_password
from ...models.user import User
from ...schemas.user import UserCreate, UserResponse
from ...core.config import settings

router = APIRouter(prefix="/auth", tags=["auth"])

@router.post("/register", response_model=UserResponse)
def register(user_data: UserCreate, db: Session = Depends(get_db)):
    existing = db.query(User).filter(User.email == user_data.email).first()
    if existing:
        raise HTTPException(400, "Email already registered")
    user = User(
        email=user_data.email,
        hashed_password=hash_password(user_data.password),
        full_name=user_data.full_name
    )
    db.add(user)
    db.commit()
    db.refresh(user)
    return user

@router.post("/login")
def login(form: OAuth2PasswordRequestForm = Depends(), db: Session = Depends(get_db)):
    user = authenticate_user(db, form.username, form.password)
    if not user:
        raise HTTPException(status.HTTP_401_UNAUTHORIZED, "Invalid credentials")
    token = create_token({"sub": user.email}, timedelta(minutes=settings.ACCESS_TOKEN_EXPIRE_MINUTES))
    return {"access_token": token, "token_type": "bearer"}"""

os.makedirs(os.path.dirname('zellaforge/backend/app/api/v1/auth.py'), exist_ok=True)
with open('zellaforge/backend/app/api/v1/auth.py', 'w') as f:
    f.write(auth_content)

print("auth.py created successfully in zellaforge/backend/app/api/v1/.")
````

````python
import os

executor_content = """import asyncio
from typing import Dict, Any
from datetime import datetime
from sqlalchemy.orm import Session
from ...models.workflow import Workflow, WorkflowRun
from ...models.task import Task
from ...core.database import get_db
from ...ai.assistant import generate_ai_response

class WorkflowExecutor:
    def __init__(self, workflow: Workflow, trigger_data: Dict):
        self.workflow = workflow
        self.trigger_data = trigger_data
        self.context = {"trigger": trigger_data, "results": {}}
        self.nodes = {n["id"]: n for n in workflow.nodes}
        self.edges = workflow.edges
        self.logs = []

    def log(self, node_id: str, status: str, msg: str, output: Any = None):
        self.logs.append({
            "node_id": node_id, "status": status, "message": msg,
            "output": output, "timestamp": datetime.utcnow().isoformat()
        })

    async def run_action(self, node: Dict) -> bool:
        action = node["data"].get("action")
        if action == "create_task":
            title = node["data"]["title"].format(**self.context["trigger"])
            db: Session = next(get_db())
            task = Task(title=title, project_id=node["data"]["project_id"], status="todo")
            db.add(task); db.commit()
            self.log(node["id"], "success", "Task created", {"task_id": task.id})
            self.context["results"][node["id"]] = {"task_id": task.id}
            return True
        return False

    async def run_ai(self, node: Dict) -> bool:
        prompt = node["data"]["prompt"].format(**self.context["trigger"])
        response = await generate_ai_response(prompt)
        self.context["results"][node["id"]] = {"response": response}
        self.log(node["id"], "success", "AI response", response[:200])
        return True

    async def run_condition(self, node: Dict) -> bool:
        cond = node["data"]["condition"]
        left = str(self.context["trigger"].get(cond["left"], ""))
        op = cond["operator"]
        right = cond["right"]

        if op == "eq":
            result = left == right
        elif op == "ne":
            result = left != right
        elif op == "contains":
            result = right in left
        else:
            result = False

        self.context["results"][node["id"]] = {"result": result}
        self.log(node["id"], "success", f"Condition evaluated: {result}", {"result": result})
        return result

    async def run(self):
        db: Session = next(get_db())
        run = WorkflowRun(workflow_id=self.workflow.id, status="running", triggered_by=self.trigger_data.get("type"))
        db.add(run); db.commit()

        try:
            incoming = {e["target"] for e in self.edges}
            queue = asyncio.Queue()
            for n in self.workflow.nodes:
                if n["id"] not in incoming:
                    await queue.put(n["id"])

            visited = set()
            while not queue.empty():
                node_id = await queue.get()
                if node_id in visited: continue
                visited.add(node_id)

                node = self.nodes[node_id]
                success = True
                if node["type"] == "action":
                    success = await self.run_action(node)
                elif node["type"] == "ai":
                    success = await self.run_ai(node)
                elif node["type"] == "condition":
                    success = await self.run_condition(node)

                if not success:
                    run.status = "failed"
                    break

                for e in self.edges:
                    if e["source"] == node_id:
                        parents = [ee["source"] for ee in self.edges if ee["target"] == e["target"]]
                        if all(p in visited for p in parents):
                            await queue.put(e["target"])

            run.status = "success" if run.status != "failed" else "failed"
        except Exception as e:
            run.status = "error"
            self.log("system", "error", str(e))

        run.logs = self.logs
        db.commit()"""

os.makedirs(os.path.dirname('zellaforge/backend/app/engine/executor.py'), exist_ok=True)
with open('zellaforge/backend/app/engine/executor.py', 'w') as f:
    f.write(executor_content)

print("executor.py created successfully in zellaforge/backend/app/engine/.")
````

````python
import os

security_content = """from datetime import datetime, timedelta, UTC
from typing import Optional
from jose import JWTError, jwt
from passlib.context import CryptContext
from fastapi import Depends, HTTPException, status
from fastapi.security import OAuth2PasswordBearer
from sqlalchemy.orm import Session
from ..models.user import User
from ..core.database import get_db
from ..core.config import settings

pwd_context = CryptContext(schemes=["bcrypt"], deprecated="auto")
oauth2_scheme = OAuth2PasswordBearer(tokenUrl="api/v1/auth/login")

def verify_password(plain: str, hashed: str) -> bool:
    return pwd_context.verify(plain, hashed)

def hash_password(password: str) -> str:
    return pwd_context.hash(password)

def create_token(data: dict, expires_delta: Optional[timedelta] = None) -> str:
    to_encode = data.copy()
    expire = datetime.now(UTC) + (expires_delta or timedelta(minutes=15))
    to_encode.update({"exp": expire})
    return jwt.encode(to_encode, settings.SECRET_KEY, algorithm=settings.ALGORITHM)

def authenticate_user(db: Session, email: str, password: str) -> Optional[User]:
    user = db.query(User).filter(User.email == email).first()
    if not user or not verify_password(password, user.hashed_password):
        return None
    return user

async def get_current_user(token: str = Depends(oauth2_scheme), db: Session = Depends(get_db)) -> User:
    try:
        payload = jwt.decode(token, settings.SECRET_KEY, algorithms=[settings.ALGORITHM])
        email: str = payload.get("sub")
        if not email:
            raise HTTPException(status_code=401, detail="Invalid token")
    except JWTError:
        raise HTTPException(status_code=401, detail="Invalid token")
    user = db.query(User).filter(User.email == email).first()
    if not user:
        raise HTTPException(status_code=401, detail="User not found")
    return user"""

os.makedirs(os.path.dirname('zellaforge/backend/app/core/security.py'), exist_ok=True)
with open('zellaforge/backend/app/core/security.py', 'w') as f:
    f.write(security_content)

print("security.py created successfully in zellaforge/backend/app/core/.")
````

````python
import os

workflow_builder_tsx_content = """import { useEffect, useState } from 'react';
import ReactFlow, { Node, Edge, addEdge, useNodesState, useEdgesState } from 'reactflow';
import 'reactflow/dist/style.css';
import api from '../services/api';
import { useParams } from 'react-router-dom';

const nodeTypes = {
  trigger: ({ data }) => <div className=\"bg-green-100 p-4 rounded border-2 border-green-500\">Trigger: {data.label}</div>,
  action: ({ data }) => <div className=\"bg-blue-100 p-4 rounded border-2 border-blue-500\">Action: {data.label}</div>,
  condition: ({ data }) => <div className=\"bg-orange-100 p-4 rounded border-2 border-orange-500\">If: {data.label}</div>,
};

export default function WorkflowBuilder() {
  const { id } = useParams();
  const [nodes, setNodes, onNodesChange] = useNodesState([]);
  const [edges, setEdges, onEdgesChange] = useEdgesState([]);

  useEffect(() => {
    if (id !== 'new') {
      api.get(`/api/v1/workflows/${id}`).then(res => {
        setNodes(res.data.nodes.map((n: any) => ({ ...n, type: n.type })));
        setEdges(res.data.edges);
      });
    }
  }, [id]);

  const onConnect = (params: any) => setEdges(eds => addEdge(params, eds));

  const onDrop = (e: React.DragEvent) => {
    e.preventDefault();
    const type = e.dataTransfer.getData('type');
    const bounds = e.currentTarget.getBoundingClientRect();
    const newNode: Node = {
      id: `${type}-${Date.now()}`,
      type,
      position: { x: e.clientX - bounds.left, y: e.clientY - bounds.top },
      data: { label: `${type} node` },
    };
    setNodes(nds => nds.concat(newNode));
  };

  const save = () => {
    const payload = { nodes, edges, name: 'My Workflow' };
    if (id === 'new') {
      api.post('/api/v1/workflows', payload);
    } else {
      api.put(`/api/v1/workflows/${id}`, payload);
    }
  };

  return (
    <div className=\"h-screen flex\">
      <div className=\"w-64 bg-gray-100 p-4\">
        <h3 className=\"font-bold mb-4\">Nodes</h3>
        {['trigger', 'action', 'condition'].map(t => (
          <div
            key={t}
            draggable
            onDragStart={e => e.dataTransfer.setData('type', t)}
            className=\"bg-white p-3 rounded shadow mb-2 cursor-grab\"
          >
            {t.charAt(0).toUpperCase() + t.slice(1)}
          </div>
        ))}
        <button onClick={save} className=\"mt-6 w-full bg-green-600 text-white p-2 rounded\">
          Save Workflow
        </button>
      </div>
      <div className=\"flex-1\" onDrop={onDrop} onDragOver={e => e.preventDefault()}>
        <ReactFlow
          nodes={nodes}
          edges={edges}
          onNodesChange={onNodesChange}
          onEdgesChange={onEdgesChange}
          onConnect={onConnect}
          nodeTypes={nodeTypes}
          fitView
        />
      </div>
    </div>
  );
}"""

os.makedirs(os.path.dirname('zellaforge/frontend/src/pages/WorkflowBuilder.tsx'), exist_ok=True)
with open('zellaforge/frontend/src/pages/WorkflowBuilder.tsx', 'w') as f:
    f.write(workflow_builder_tsx_content)

print("WorkflowBuilder.tsx created successfully in zellaforge/frontend/src/pages/.")
````

````python
import os

executor_content = """import asyncio
from typing import Dict, Any
from datetime import datetime
from sqlalchemy.orm import Session
from ...models.workflow import Workflow, WorkflowRun
from ...models.task import Task
from ...core.database import get_db
from ...ai.assistant import generate_ai_response

class WorkflowExecutor:
    def __init__(self, workflow: Workflow, trigger_data: Dict):
        self.workflow = workflow
        self.trigger_data = trigger_data
        self.context = {"trigger": trigger_data, "results": {}}
        self.nodes = {n["id"]: n for n in workflow.nodes}
        self.edges = workflow.edges
        self.logs = []

    def log(self, node_id: str, status: str, msg: str, output: Any = None):
        self.logs.append({
            "node_id": node_id, "status": status, "message": msg,
            "output": output, "timestamp": datetime.utcnow().isoformat()
        })

    async def run_action(self, node: Dict) -> bool:
        action = node["data"].get("action")
        if action == "create_task":
            title = node["data"]["title"].format(**self.context["trigger"])
            db: Session = next(get_db())
            task = Task(title=title, project_id=node["data"]["project_id"], status="todo")
            db.add(task); db.commit()
            self.log(node["id"], "success", "Task created", {"task_id": task.id})
            self.context["results"][node["id"]] = {"task_id": task.id}
            return True
        return False

    async def run_ai(self, node: Dict) -> bool:
        prompt = node["data"]["prompt"].format(**self.context["trigger"])
        response = await generate_ai_response(prompt)
        self.context["results"][node["id"]] = {"response": response}
        self.log(node["id"], "success", "AI response", response[:200])
        return True

    async def run_condition(self, node: Dict) -> bool:
        cond = node["data"]["condition"]
        left = str(self.context["trigger"].get(cond["left"], ""))
        op = cond["operator"]
        right = cond["right"]

        if op == "eq":
            result = left == right
        elif op == "ne":
            result = left != right
        elif op == "contains":
            result = right in left
        else:
            result = False

        self.context["results"][node["id"]] = {"result": result}
        self.log(node["id"], "success", f"Condition evaluated: {result}", {"result": result})
        return result

    async def run(self):
        db: Session = next(get_db())
        run = WorkflowRun(workflow_id=self.workflow.id, status="running", triggered_by=self.trigger_data.get("type"))
        db.add(run); db.commit()

        try:
            incoming = {e["target"] for e in self.edges}
            queue = asyncio.Queue()
            for n in self.workflow.nodes:
                if n["id"] not in incoming:
                    await queue.put(n["id"])

            visited = set()
            while not queue.empty():
                node_id = await queue.get()
                if node_id in visited: continue
                visited.add(node_id)

                node = self.nodes[node_id]
                success = True
                if node["type"] == "action":
                    success = await self.run_action(node)
                elif node["type"] == "ai":
                    success = await self.run_ai(node)
                elif node["type"] == "condition":
                    success = await self.run_condition(node)

                if not success:
                    run.status = "failed"
                    break

                for e in self.edges:
                    if e["source"] == node_id:
                        parents = [ee["source"] for ee in self.edges if ee["target"] == e["target"]]
                        if all(p in visited for p in parents):
                            await queue.put(e["target"])

            run.status = "success" if run.status != "failed" else "failed"
        except Exception as e:
            run.status = "error"
            self.log("system", "error", str(e))

        run.logs = self.logs
        db.commit()"""

os.makedirs(os.path.dirname('zellaforge/backend/app/engine/executor.py'), exist_ok=True)
with open('zellaforge/backend/app/engine/executor.py', 'w') as f:
    f.write(executor_content)

print("executor.py created successfully in zellaforge/backend/app/engine/.")
````

````python
import os

workflow_api_content = """from fastapi import APIRouter, Depends, HTTPException
from sqlalchemy.orm import Session
from typing import List
from ...core.database import get_db
from ...core.security import get_current_user
from ...models.workflow import Workflow
from ...schemas.workflow import WorkflowCreate, WorkflowUpdate, WorkflowResponse

router = APIRouter(prefix="/workflows", tags=["workflows"])

@router.post("/", response_model=WorkflowResponse)
def create_workflow(
    data: WorkflowCreate,
    db: Session = Depends(get_db),
    user=Depends(get_current_user)
):
    wf = Workflow(
        name=data.name,
        description=data.description,
        owner_id=user.id,
        nodes=[n.dict() for n in data.nodes],
        edges=[e.dict() for e in data.edges]
    )
    db.add(wf)
    db.commit()
    db.refresh(wf)
    return wf

@router.get("/", response_model=List[WorkflowResponse])
def list_workflows(db: Session = Depends(get_db), user=Depends(get_current_user)):
    return db.query(Workflow).filter(Workflow.owner_id == user.id).all()

@router.put("/{wf_id}", response_model=WorkflowResponse)
def update_workflow(
    wf_id: int,
    data: WorkflowUpdate,
    db: Session = Depends(get_db),
    user=Depends(get_current_user)
):
    wf = db.query(Workflow).filter(Workflow.id == wf_id, Workflow.owner_id == user.id).first()
    if not wf:
        raise HTTPException(404, "Not found")
    if data.name: wf.name = data.name
    if data.nodes: wf.nodes = [n.dict() for n in data.nodes]
    if data.edges: wf.edges = [e.dict() for e in data.edges]
    if data.is_active is not None: wf.is_active = data.is_active
    db.commit()
    db.refresh(wf)
    return wf"""

os.makedirs(os.path.dirname('zellaforge/backend/app/api/v1/workflows.py'), exist_ok=True)
with open('zellaforge/backend/app/api/v1/workflows.py', 'w') as f:
    f.write(workflow_api_content)

print("workflows.py created successfully in zellaforge/backend/app/api/v1/.")
````

import os

executor_content = """import asyncio
from typing import Dict, Any
from datetime import datetime
from sqlalchemy.orm import Session
from app.models.workflow import Workflow, WorkflowRun
from app.models.task import Task
from app.core.database import get_db
from app.ai.assistant import generate_ai_response

class WorkflowExecutor:
    def __init__(self, workflow: Workflow, trigger_data: Dict):
        self.workflow = workflow
        self.trigger_data = trigger_data
        self.context = {"trigger": trigger_data, "results": {}}
        self.nodes = {n["id"]: n for n in workflow.nodes}
        self.edges = workflow.edges
        self.logs = []

    def log(self, node_id: str, status: str, msg: str, output: Any = None):
        self.logs.append({
            "node_id": node_id, "status": status, "message": msg,
            "output": output, "timestamp": datetime.utcnow().isoformat()
        })

    async def run_action(self, node: Dict) -> bool:
        action = node["data"].get("action")
        if action == "create_task":
            title = node["data"]["title"].format(**self.context["trigger"])
            db: Session = next(get_db())
            task = Task(title=title, project_id=node["data"]["project_id"], status="todo")
            db.add(task); db.commit()
            self.log(node["id"], "success", "Task created", {"task_id": task.id})
            self.context["results"][node["id"]] = {"task_id": task.id}
            return True
        return False

````python
import os

workflow_builder_tsx_content = """import { useEffect, useState } from 'react';
import ReactFlow, { Node, Edge, addEdge, useNodesState, useEdgesState } from 'reactflow';
import 'reactflow/dist/style.css';
import api from '../services/api';
import { useParams } from 'react-router-dom';

const nodeTypes = {
  trigger: ({ data }) => <div className=\"bg-green-100 p-4 rounded border-2 border-green-500\">Trigger: {data.label}</div>,
  action: ({ data }) => <div className=\"bg-blue-100 p-4 rounded border-2 border-blue-500\">Action: {data.label}</div>,
  condition: ({ data }) => <div className=\"bg-orange-100 p-4 rounded border-2 border-orange-500\">If: {data.label}</div>,
};

export default function WorkflowBuilder() {
  const { id } = useParams();
  const [nodes, setNodes, onNodesChange] = useNodesState([]);
  const [edges, setEdges, onEdgesChange] = useEdgesState([]);

  useEffect(() => {
    if (id !== 'new') {
      api.get(`/api/v1/workflows/${id}`).then(res => {
        setNodes(res.data.nodes.map((n: any) => ({ ...n, type: n.type })));
        setEdges(res.data.edges);
      });
    }
  }, [id]);

  const onConnect = (params: any) => setEdges(eds => addEdge(params, eds));

  const onDrop = (e: React.DragEvent) => {
    e.preventDefault();
    const type = e.dataTransfer.getData('type');
    const bounds = e.currentTarget.getBoundingClientRect();
    const newNode: Node = {
      id: `${type}-${Date.now()}`,
      type,
      position: { x: e.clientX - bounds.left, y: e.clientY - bounds.top },
      data: { label: `${type} node` },
    };
    setNodes(nds => nds.concat(newNode));
  };

  const save = () => {
    const payload = { nodes, edges, name: 'My Workflow' };
    if (id === 'new') {
      api.post('/api/v1/workflows', payload);
    } else {
      api.put(`/api/v1/workflows/${id}`, payload);
    }
  };

  return (
    <div className=\"h-screen flex\">
      <div className=\"w-64 bg-gray-100 p-4\">
        <h3 className=\"font-bold mb-4\">Nodes</h3>
        {['trigger', 'action', 'condition'].map(t => (
          <div
            key={t}
            draggable
            onDragStart={e => e.dataTransfer.setData('type', t)}
            className=\"bg-white p-3 rounded shadow mb-2 cursor-grab\"
          >
            {t.charAt(0).toUpperCase() + t.slice(1)}
          </div>
        ))}
        <button onClick={save} className=\"mt-6 w-full bg-green-600 text-white p-2 rounded\">
          Save Workflow
        </button>
      </div>
      <div className=\"flex-1\" onDrop={onDrop} onDragOver={e => e.preventDefault()}>
        <ReactFlow
          nodes={nodes}
          edges={edges}
          onNodesChange={onNodesChange}
          onEdgesChange={onEdgesChange}
          onConnect={onConnect}
          nodeTypes={nodeTypes}
          fitView
        />
      </div>
    </div>
  );
}"""

os.makedirs(os.path.dirname('zellaforge/frontend/src/pages/WorkflowBuilder.tsx'), exist_ok=True)
with open('zellaforge/frontend/src/pages/WorkflowBuilder.tsx', 'w') as f:
    f.write(workflow_builder_tsx_content)

print("WorkflowBuilder.tsx created successfully in zellaforge/frontend/src/pages/.")
````
